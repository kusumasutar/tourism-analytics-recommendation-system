```python
# -----------------------------------------
# Step 1: Import required libraries
# -----------------------------------------

import pandas as pd
import numpy as np

from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

#### Load the cleaned dataset

```python
# -----------------------------------------
# Step 2: Load cleaned tourism data
# -----------------------------------------

df = pd.read_csv("../data/processed/cleaned_data.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

# Display the first 5 rows
df.head()
```

#### Build the User-Items Ratings Matrix

```python
# -----------------------------------------
# Step 3: Build User-Item Ratings Matrix
# -----------------------------------------

# Each row represents a user
# Each column represents an attraction
# Each value represents the rating given by the user

user_item_matrix = df.pivot_table(
    index="UserId",
    columns="AttractionId",
    values="Rating",
    aggfunc="mean",
    fill_value=0
)

print("User-Item Matrix created successfully!")

print("Matrix shape:", user_item_matrix.shape)

# Show the first 5 users
user_item_matrix.head()
```

### Calculate Attraction Similarity

```python
# -----------------------------------------
# Step 4: Calculate Attraction Similarity
# -----------------------------------------

# Transpose the matrix:
# Rows will become attractions
# Columns will become users
#
# This allows us to compare attractions
# based on the ratings given by users.

item_user_matrix = user_item_matrix.T

# Calculate cosine similarity between attractions
item_similarity = cosine_similarity(item_user_matrix)

# Convert the similarity array into a DataFrame
# so that we can easily identify attractions by ID.

item_similarity_df = pd.DataFrame(
    item_similarity,
    index=item_user_matrix.index,
    columns=item_user_matrix.index
)

print("Attraction similarity calculated successfully!")

print("Similarity matrix shape:", item_similarity_df.shape)

# Display the first 5 rows
item_similarity_df.head()
```

#### Create Attraction Information table

```python
# -----------------------------------------
# Step 5: Create Attraction Information
# -----------------------------------------

# We only need one row for each attraction.
# drop_duplicates() prevents the same attraction
# from appearing multiple times.

attraction_info = df[
    [
        "AttractionId",
        "Attraction",
        "AttractionType",
        "AttractionCityId",
        "CityName",
        "Country",
        "Region",
        "Continent"
    ]
].drop_duplicates("AttractionId").copy()

print("Attraction information created successfully!")

print("Number of attractions:", len(attraction_info))

attraction_info.head()
```

#### Create a Collaborative Recommendation Function

```python
# -----------------------------------------
# Step 6: Collaborative Filtering Function
# -----------------------------------------

def collaborative_recommend(user_id, n=5):
    """
    Recommend attractions using item-based
    collaborative filtering.

    Parameters:
        user_id : ID of the user
        n       : number of recommendations

    Returns:
        DataFrame containing recommended attractions
    """

    # Check whether the user exists in our matrix
    if user_id not in user_item_matrix.index:
        return pd.DataFrame()

    # Get ratings given by this user
    user_ratings = user_item_matrix.loc[user_id]

    # Keep only attractions that the user has actually rated
    rated_items = user_ratings[user_ratings > 0]

    # If the user has no ratings, collaborative filtering
    # cannot determine their preferences.
    if len(rated_items) == 0:
        return pd.DataFrame()

    # Calculate recommendation scores.
    #
    # Each rated attraction contributes to the score
    # according to:
    #
    # user rating × similarity to another attraction
    #
    recommendation_scores = item_similarity_df[rated_items.index].dot(
        rated_items
    )

    # Calculate the total similarity weight.
    similarity_weights = item_similarity_df[rated_items.index].sum(axis=1)

    # Avoid division by zero
    similarity_weights = similarity_weights.replace(0, np.nan)

    # Normalize the score
    recommendation_scores = (
        recommendation_scores / similarity_weights
    ).fillna(0)

    # Do not recommend attractions that the user
    # has already rated.
    recommendation_scores = recommendation_scores.drop(
        labels=rated_items.index,
        errors="ignore"
    )

    # Select the highest-scoring attractions
    recommendations = recommendation_scores.sort_values(
        ascending=False
    ).head(n)

    # Convert the Series into a DataFrame
    result = recommendations.reset_index()

    result.columns = [
        "AttractionId",
        "RecommendationScore"
    ]

    # Add attraction details
    result = result.merge(
        attraction_info,
        on="AttractionId",
        how="left"
    )

    return result
```

### Test Collaborative Filtering

```python
# -----------------------------------------
# Step 7: Test Collaborative Filtering
# -----------------------------------------

# Select the first user available in the matrix
test_user_id = user_item_matrix.index[0]

print("Testing recommendation for UserId:", test_user_id)

collaborative_recommend(
    user_id=test_user_id,
    n=5
)
```

#### Build the Content-Based Recommendation System

Collaborative filtering works well when we have user history.

But what about a new user?

For example:

"I like nature attractions and want somewhere in Europe."

There may be no previous ratings for that user.

So we create a second recommendation method based on:

Attraction type
City
Country
Region
Continent

```python
# -----------------------------------------
# Step 8: Prepare Content-Based Features
# -----------------------------------------

# Create a copy so we don't accidentally modify
# the original attraction information table.

content_data = attraction_info.copy()

# Replace missing values with "Unknown"
content_columns = [
    "AttractionType",
    "CityName",
    "Country",
    "Region",
    "Continent"
]

for column in content_columns:
    content_data[column] = (
        content_data[column]
        .fillna("Unknown")
        .astype(str)
    )

# Combine the important descriptive information
# into one text representation.

content_data["Content"] = (
    content_data["AttractionType"] + " "
    + content_data["CityName"] + " "
    + content_data["Country"] + " "
    + content_data["Region"] + " "
    + content_data["Continent"]
)

print("Content features prepared successfully!")

content_data[
    ["AttractionId", "Attraction", "Content"]
].head()
```

#### Calculate Content Similarity

```python
# -----------------------------------------
# Step 9: Import TF-IDF
# -----------------------------------------

from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# -----------------------------------------
# Create TF-IDF representation
# -----------------------------------------

tfidf = TfidfVectorizer()

content_matrix = tfidf.fit_transform(
    content_data["Content"]
)

# Calculate similarity between attractions
content_similarity = cosine_similarity(
    content_matrix
)

content_similarity_df = pd.DataFrame(
    content_similarity,
    index=content_data["AttractionId"],
    columns=content_data["AttractionId"]
)

print("Content-based similarity calculated successfully!")

print(
    "Content similarity shape:",
    content_similarity_df.shape
)
```

#### Content-Based Recommendation Function

```python
# -----------------------------------------
# Step 10: Content-Based Recommendation
# -----------------------------------------

def content_recommend(
    attraction_type=None,
    continent=None,
    region=None,
    n=5
):
    """
    Recommend attractions based on attraction type
    and location preferences.
    """

    scores = pd.Series(
        0.0,
        index=content_data["AttractionId"]
    )

    # Attraction type preference
    if attraction_type:
        matching_ids = content_data.loc[
            content_data["AttractionType"].str.lower()
            == str(attraction_type).lower(),
            "AttractionId"
        ]

        scores.loc[scores.index.isin(matching_ids)] += 3

    # Continent preference
    if continent:
        matching_ids = content_data.loc[
            content_data["Continent"].str.lower()
            == str(continent).lower(),
            "AttractionId"
        ]

        scores.loc[scores.index.isin(matching_ids)] += 2

    # Region preference
    if region:
        matching_ids = content_data.loc[
            content_data["Region"].str.lower()
            == str(region).lower(),
            "AttractionId"
        ]

        scores.loc[scores.index.isin(matching_ids)] += 2

    # Sort by preference score
    scores = scores.sort_values(
        ascending=False
    ).head(n)

    result = scores.reset_index()

    result.columns = [
        "AttractionId",
        "ContentScore"
    ]

    result = result.merge(
        attraction_info,
        on="AttractionId",
        how="left"
    )

    return result
```

#### Test Content-Based Recommendation

```python
# -----------------------------------------
# Step 11: Test Content-Based Recommendation
# -----------------------------------------

content_recommend(
    attraction_type="Nature & Wildlife Areas",
    continent="Europe",
    n=5
)
```

#### Create the Final recommend() Function

```python
# -----------------------------------------
# Step 12: Final Recommendation Function
# -----------------------------------------

def recommend(
    user_id=None,
    n=5,
    attraction_type=None,
    continent=None,
    region=None
):
    """
    Main recommendation function.

    For existing users:
        Uses collaborative filtering.

    For new users:
        Uses content-based filtering.

    If preferences are supplied, they are used
    to improve the recommendation ranking.
    """

    # -------------------------------------
    # 1. Try collaborative recommendations
    # -------------------------------------

    collaborative_result = pd.DataFrame()

    if user_id is not None:
        collaborative_result = collaborative_recommend(
            user_id=user_id,
            n=30
        )

    # -------------------------------------
    # 2. Get content-based recommendations
    # -------------------------------------

    content_result = content_recommend(
        attraction_type=attraction_type,
        continent=continent,
        region=region,
        n=30
    )

    # -------------------------------------
    # 3. Combine the two recommendation
    #    approaches
    # -------------------------------------

    if not collaborative_result.empty:

        final_result = collaborative_result[
            [
                "AttractionId",
                "RecommendationScore",
                "Attraction",
                "AttractionType",
                "CityName",
                "Country",
                "Region",
                "Continent"
            ]
        ].copy()

        # Add content score where available
        if not content_result.empty:

            content_scores = content_result[
                [
                    "AttractionId",
                    "ContentScore"
                ]
            ]

            final_result = final_result.merge(
                content_scores,
                on="AttractionId",
                how="left"
            )

        else:
            final_result["ContentScore"] = 0

        final_result["ContentScore"] = (
            final_result["ContentScore"].fillna(0)
        )

        # Normalize collaborative score
        max_collab = final_result[
            "RecommendationScore"
        ].max()

        if max_collab > 0:
            final_result["CollaborativeScore"] = (
                final_result["RecommendationScore"]
                / max_collab
            )
        else:
            final_result["CollaborativeScore"] = 0

        # Combine both approaches
        final_result["FinalScore"] = (
            0.7 * final_result["CollaborativeScore"]
            + 0.3 * final_result["ContentScore"]
        )

    else:

        # No previous user history.
        # Therefore use content-based recommendations.

        final_result = content_result.copy()

        if final_result.empty:
            return pd.DataFrame()

        final_result["CollaborativeScore"] = 0
        final_result["RecommendationScore"] = 0

        # Content score is already our main score
        final_result["FinalScore"] = (
            final_result["ContentScore"]
        )

    # Sort by final recommendation score
    final_result = final_result.sort_values(
        "FinalScore",
        ascending=False
    )

    # Return top N recommendations
    return final_result.head(n).reset_index(drop=True)
```

```python
# -----------------------------------------
# Test with a new user
# -----------------------------------------

recommend(
    user_id=None,
    n=5,
    attraction_type="Nature & Wildlife Areas",
    continent="Europe"
)
```
