## Build the final cleaned dataset

```python
# ============================================================
# STEP 4: CLEAN, MERGE AND CREATE FINAL DATASET
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STARTING FINAL DATA CLEANING AND MERGING")
print("=" * 70)

# ------------------------------------------------------------
# 1. Make copies so original raw tables remain unchanged
# ------------------------------------------------------------

tx = transaction.copy()
usr = user.copy()
cty = city.copy()
cntry = country.copy()
reg = region.copy()
cont = continent.copy()
itm = item.copy()
typ = type_.copy()
md = mode.copy()

print("Original transaction rows:", len(tx))


# ------------------------------------------------------------
# 2. Standardize text fields
# ------------------------------------------------------------

text_columns = {
    "Country": cntry,
    "Region": reg,
    "Continent": cont,
    "CityName": cty,
    "Attraction": itm,
    "AttractionAddress": itm,
    "AttractionType": typ,
    "VisitMode": md
}

for column_name, df in text_columns.items():
    if column_name in df.columns:
        df[column_name] = (
            df[column_name]
            .astype("string")
            .str.strip()
        )

print("\nText fields standardized:")
print("- Leading/trailing whitespace removed")
print("- Original capitalization preserved")


# ------------------------------------------------------------
# 3. Convert placeholder "-" to missing values
# ------------------------------------------------------------

for df in [cntry, reg, cont, cty, typ, md]:
    for col in df.columns:
        if df[col].dtype == "string":
            df[col] = df[col].replace("-", pd.NA)

print("\nPlaceholder '-' converted to missing values where applicable.")


# ------------------------------------------------------------
# 4. Validate lookup-table keys before merging
# ------------------------------------------------------------

key_checks = {
    "User.UserId": (usr, "UserId"),
    "City.CityId": (cty, "CityId"),
    "Country.CountryId": (cntry, "CountryId"),
    "Region.RegionId": (reg, "RegionId"),
    "Continent.ContinentId": (cont, "ContinentId"),
    "Item.AttractionId": (itm, "AttractionId"),
    "Type.AttractionTypeId": (typ, "AttractionTypeId"),
    "Mode.VisitModeId": (md, "VisitModeId")
}

print("\n" + "=" * 70)
print("PRIMARY KEY VALIDATION")
print("=" * 70)

for name, (df, key) in key_checks.items():
    duplicate_keys = df[key].duplicated().sum()
    print(f"{name}: {duplicate_keys} duplicate keys")

    if duplicate_keys > 0:
        raise ValueError(
            f"{name} contains duplicate keys. "
            "Fix this before merging to avoid multiplying transaction rows."
        )

print("All lookup keys are unique.")


# ------------------------------------------------------------
# 5. Rename Transaction.VisitMode to VisitModeId
# ------------------------------------------------------------

tx = tx.rename(columns={"VisitMode": "VisitModeId"})


# ------------------------------------------------------------
# 6. Create VisitDate
# ------------------------------------------------------------

tx["VisitDate"] = pd.to_datetime(
    {
        "year": tx["VisitYear"],
        "month": tx["VisitMonth"],
        "day": 1
    },
    errors="coerce"
)

invalid_dates = tx["VisitDate"].isna().sum()

print("\n" + "=" * 70)
print("DATE CREATION")
print("=" * 70)
print("Invalid VisitDate values:", invalid_dates)

if invalid_dates > 0:
    raise ValueError("Some VisitYear/VisitMonth combinations produced invalid dates.")

print("VisitDate created successfully.")


# ------------------------------------------------------------
# 7. Merge User information
# ------------------------------------------------------------

user_columns = [
    "UserId",
    "ContinentId",
    "RegionId",
    "CountryId",
    "CityId"
]

final_df = tx.merge(
    usr[user_columns],
    on="UserId",
    how="left",
    validate="many_to_one"
)

print("\nAfter User merge:", final_df.shape)


# ------------------------------------------------------------
# 8. Merge City information
# ------------------------------------------------------------

city_lookup = cty[
    ["CityId", "CityName"]
].copy()

final_df = final_df.merge(
    city_lookup,
    on="CityId",
    how="left",
    validate="many_to_one"
)

print("After City merge:", final_df.shape)


# ------------------------------------------------------------
# 9. Merge Country information
# ------------------------------------------------------------

country_lookup = cntry[
    ["CountryId", "Country"]
].copy()

final_df = final_df.merge(
    country_lookup,
    on="CountryId",
    how="left",
    validate="many_to_one"
)

print("After Country merge:", final_df.shape)


# ------------------------------------------------------------
# 10. Merge Region information
# ------------------------------------------------------------

region_lookup = reg[
    ["RegionId", "Region"]
].copy()

final_df = final_df.merge(
    region_lookup,
    on="RegionId",
    how="left",
    validate="many_to_one"
)

print("After Region merge:", final_df.shape)


# ------------------------------------------------------------
# 11. Merge Continent information
# ------------------------------------------------------------

continent_lookup = cont[
    ["ContinentId", "Continent"]
].copy()

final_df = final_df.merge(
    continent_lookup,
    on="ContinentId",
    how="left",
    validate="many_to_one"
)

print("After Continent merge:", final_df.shape)


# ------------------------------------------------------------
# 12. Merge Item information
# ------------------------------------------------------------

item_columns = [
    "AttractionId",
    "AttractionCityId",
    "AttractionTypeId",
    "Attraction",
    "AttractionAddress"
]

final_df = final_df.merge(
    itm[item_columns],
    on="AttractionId",
    how="left",
    validate="many_to_one"
)

print("After Item merge:", final_df.shape)


# ------------------------------------------------------------
# 13. Merge Attraction Type
# ------------------------------------------------------------

type_lookup = typ[
    ["AttractionTypeId", "AttractionType"]
].copy()

final_df = final_df.merge(
    type_lookup,
    on="AttractionTypeId",
    how="left",
    validate="many_to_one"
)

print("After Attraction Type merge:", final_df.shape)


# ------------------------------------------------------------
# 14. Merge Visit Mode description
# ------------------------------------------------------------

mode_lookup = md[
    ["VisitModeId", "VisitMode"]
].copy()

final_df = final_df.merge(
    mode_lookup,
    on="VisitModeId",
    how="left",
    validate="many_to_one"
)

print("After Visit Mode merge:", final_df.shape)


# ------------------------------------------------------------
# 15. Reorder columns
# ------------------------------------------------------------

desired_columns = [
    "TransactionId",
    "UserId",

    "VisitDate",
    "VisitYear",
    "VisitMonth",

    "VisitModeId",
    "VisitMode",

    "AttractionId",
    "Attraction",
    "AttractionAddress",
    "AttractionCityId",
    "AttractionTypeId",
    "AttractionType",

    "Rating",

    "CityId",
    "CityName",
    "CountryId",
    "Country",
    "RegionId",
    "Region",
    "ContinentId",
    "Continent"
]

# Keep only columns that exist
desired_columns = [
    col for col in desired_columns
    if col in final_df.columns
]

final_df = final_df[desired_columns]


# ------------------------------------------------------------
# 16. Final data types
# ------------------------------------------------------------

final_df["VisitDate"] = pd.to_datetime(final_df["VisitDate"])

# Keep IDs as nullable integers because CityId contains missing values
for col in [
    "UserId",
    "VisitModeId",
    "AttractionId",
    "AttractionCityId",
    "AttractionTypeId",
    "CityId",
    "CountryId",
    "RegionId",
    "ContinentId"
]:
    if col in final_df.columns:
        final_df[col] = final_df[col].astype("Int64")


# ------------------------------------------------------------
# 17. Final validation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL VALIDATION")
print("=" * 70)

print("Final dataset shape:", final_df.shape)

print("\nOriginal transaction rows:", len(transaction))
print("Final dataset rows:", len(final_df))

if len(final_df) != len(transaction):
    raise ValueError(
        "Row count changed! Check the merge logic before saving."
    )

print("✓ Transaction row count preserved.")


# Check duplicate transaction IDs
duplicate_transactions = final_df["TransactionId"].duplicated().sum()

print("\nDuplicate TransactionIds:", duplicate_transactions)

if duplicate_transactions > 0:
    raise ValueError("Duplicate TransactionIds detected.")

print("✓ TransactionId remains unique.")


# Check missing important descriptive fields
print("\nMissing values in important columns:")

important_columns = [
    "TransactionId",
    "UserId",
    "VisitDate",
    "VisitMode",
    "Attraction",
    "AttractionType",
    "Rating",
    "Country",
    "Region",
    "Continent"
]

for col in important_columns:
    print(f"{col}: {final_df[col].isna().sum()}")


# Rating validation
invalid_ratings = final_df[
    ~final_df["Rating"].between(1, 5)
]

print("\nInvalid ratings:", len(invalid_ratings))

if len(invalid_ratings) > 0:
    raise ValueError("Invalid ratings detected.")


# Date validation
print(
    "VisitDate range:",
    final_df["VisitDate"].min(),
    "to",
    final_df["VisitDate"].max()
)


# ------------------------------------------------------------
# 18. Preview
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL DATASET PREVIEW")
print("=" * 70)

display(final_df.head(10))

print("\nFinal columns:")
print(final_df.columns.tolist())

print("\nFinal dataset info:")
final_df.info()
```

What this cell is doing

The important thing is that we're using left joins throughout.

That means:

Every one of the original 52,930 transactions must remain.

We're not deleting transactions just because some user/city information is missing.

For the four users without CityId, their transactions will remain, but CityId and CityName will be missing. Their Country, Region and Continent are still retained from the User table.
